# AML Anomaly Detection using Behavioral Analysis

## Objective

The objective of this project is to identify suspicious transactions using anomaly detection techniques based on transaction amount, behavioral patterns, deviation from normal activity, and transaction frequency.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)

df.head()
```

## Data Understanding

- total_bill → transaction amount  
- tip → additional behavior indicator  
- time → transaction context (Lunch/Dinner)  
- size → used to simulate user/group behavior  

## Step 1: High Transaction Detection

Transactions with unusually high amounts are considered suspicious.

```python
df['high_amount'] = df['total_bill'] > 25
```

## Step 2: Behavioral Pattern Analysis

Transaction context such as time (Dinner/Lunch) is used to identify unusual behavior.

```python
df['is_dinner'] = df['time'] == 'Dinner'
```

## Step 3: User Behavior Analysis

Average transaction value is calculated to understand normal behavior.

```python
df['user_avg'] = df.groupby('size')['total_bill'].transform('mean')
```

## Step 4: Deviation Detection

Transactions significantly higher than the average behavior are flagged.

```python
df['deviation'] = df['total_bill'] > (df['user_avg'] * 1.2)
```

## Step 5: Frequency Analysis

High number of transactions in a group indicates potential suspicious activity.

```python
df['txn_count'] = df.groupby('size')['total_bill'].transform('count')
df['high_freq'] = df['txn_count'] > 5
```

## Step 6: Final AML Detection

Transactions are flagged if they meet one or more anomaly conditions.

```python
df['aml_flag'] = (
    df['high_amount'] |
    df['deviation'] |
    df['high_freq']
)
```

```python
df['aml_flag'].value_counts()
df[df['aml_flag'] == True].head()
```

## Note on Output and Implementation

In this project, several steps such as feature creation (e.g., high_amount, deviation, and frequency flags) do not produce direct output. This is because these steps are used to build intermediate variables required for anomaly detection.

The final output of the system is represented through the 'aml_flag' column, which combines multiple detection conditions to identify suspicious transactions.

This approach reflects real-world AML systems, where multiple internal calculations are performed before generating a final decision or alert.

- Intermediate steps focus on data transformation, while the final result focuses on decision-making.

## Result

The system successfully identifies suspicious transactions by combining multiple anomaly detection techniques, reducing false positives and improving accuracy.

## Conclusion

This project demonstrates how AML anomaly detection can be implemented using rule-based logic. By combining transaction amount, behavior, deviation, and frequency, the system can effectively detect suspicious activity similar to real-world financial systems.
